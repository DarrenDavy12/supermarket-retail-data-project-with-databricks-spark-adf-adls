# Bronze Layer - Extract 

### Get raw data from Data Lake 'bronze' container and read data to dataframe using Pyspark 

```python
# Configuration
# NOTE: In production, these values would be retrieved from Azure Key Vault
# via Databricks secret scopes (dbutils.secrets.get) to avoid hardcoding.
# Example: storage_account = dbutils.secrets.get(scope="adls-scope", key="storage-account-name")
storage_account = "datalakeretail01"
```

```python
# Import PySpark functions

from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()


# Read Bronze layer 

bronze_path = "abfss://bronze@<storage-account>.dfs.core.windows.net/supermarket_sales_raw.csv" # path of bronze container in Azure Data Lake 

df_bronze = spark.read.option("header", True).option("inferSchema", True).csv(bronze_path) 

df_bronze.show(5) # show first 5 rows 
```
